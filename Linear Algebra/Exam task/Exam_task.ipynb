$$ \frac{2^{(2n)} \cdot (2^2)^{(n+1)}}{(2^3)^{(n-1)} \cdot (2^4)^n} $$

$$ (2^2)^{(n+1)} = 2^{2(n+1)} = 2^{(2n + 2)} $$

$$ (2^4)^n = 2^{4n} $$

Первая дробь: $$ \frac{5x - 2}{3} = \frac{4(5x - 2)}{12} = \frac{20x - 8}{12} $$
Вторая дробь: $$ \frac{2x + 4}{2} = \frac{6(2x + 4)}{12} = \frac{12x + 24}{12} $$
Третья дробь: $$ \frac{7x - 5}{4} = \frac{3(7x - 5)}{12} = \frac{21x - 15}{12} $$
Четвертая дробь: $$ \frac{x + 6}{3} = \frac{4(x + 6)}{12} = \frac{4x + 24}{12} $$

Теперь подставим эти преобразования в неравенство:

$$ \frac{20x - 8}{12} - \frac{12x + 24}{12} > \frac{21x - 15}{12} + \frac{4x + 24}{12} $$

Шаг 2: Упростим неравенство
Теперь, когда у нас все дроби имеют одинаковый знаменатель (12), мы можем упростить неравенство, убрав общий знаменатель:

$$ \frac{(20x - 8) - (12x + 24)}{12} > \frac{(21x - 15) + (4x + 24)}{12} $$

После этого можем избавиться от дробей:

$$ (20x - 8) - (12x + 24) > (21x - 15) + (4x + 24) $$

Теперь упростим обе части неравенства.

Левая часть:
$$ 20x - 8 - 12x - 24 = (20x - 12x) + (-8 - 24) = 8x - 32 $$

Правая часть:
$$ 21x - 15 + 4x + 24 = (21x + 4x) + (-15 + 24) = 25x + 9 $$

Теперь наши преобразования привели к следующему неравенству:

$$ 8x - 32 > 25x + 9 $$

Шаг 3: Переносим все члены с x в одну сторону

Теперь мы перенесем все члены с  на одну сторону, а свободные члены на другую:

$$ 8x - 25x > 9 + 32 $$

Это даст нам:

$$ -17x > 41 $$

Шаг 4: Разделим на -17 и изменим знак неравенства
Чтобы найти 
, нам нужно разделить обе стороны на -17. Однако, при делении на отрицательное число неравенство меняет знак:

$$ x < -\frac{41}{17} $$

Шаг 5: Заключение
Таким образом, мы пришли кsolution нашего неравенства:

$$ x < -\frac{41}{17} $$

### Решение на Python

```python
import sympy as sp

# Объявляем переменную
x = sp.symbols('x')

# Задаем неравенство
inequality = (5*x - 2)/3 - (2*x + 4)/2 > (7*x - 5)/4 + (x + 6)/3

# Решаем неравенство
solution = sp.solve(inequality, x)

# Выводим решение
print("Решение неравенства:", solution)
```

- B слева от A

```python
import math

# Шаг 1: Определяем общее количество книг
n = 5

# Шаг 2: Определяем количество книг, которые должны стоять рядом
# В данном случае 2 книги будем считать суперкнигой
superbook_count = n - 1  # 5 - 1 = 4
ways_to_arrange_superbooks = math.factorial(superbook_count)

# Шаг 3: Способы расстановки двух книг внутри суперкниги
ways_to_arrange_A_and_B = 2  # A слева, B справа или B слева, A справа

# Шаг 4: Общее количество способов расставить книги
total_ways = ways_to_arrange_superbooks * ways_to_arrange_A_and_B

print(f"Существует {total_ways} способов расставить 5 различных книг на полке, если две из них должны стоять рядом.")
```

Начнем с определения вероятностей:

1. Вероятность, что первый телефон сломан:

2. Вероятность, что второй телефон сломан:
$$ P(B) = 0.3 $$

Теперь найдем вероятность того, что каждый из телефонов работает.

3. Вероятность, что первый телефон работает:
$$ P(A') = 1 - P(A) = 1 - 0.1 = 0.9 $$

4. Вероятность, что второй телефон работает:
$$ P(B') = 1 - P(B) = 1 - 0.3 = 0.7 $$

Теперь, чтобы найти вероятность того, что оба телефона работают одновременно, предположим, что события "работоспособность первого телефона" и "работоспособность второго телефона" независимы. Тогда вероятность того, что оба телефона работают, вычисляется как произведение их вероятностей работоспособности:

5. Вероятность, что оба телефона работают:
$$ P(A' \cap B') = P(A') \cdot P(B') = 0.9 \cdot 0.7 $$

Теперь подставим значения:

$$ P(A' \cap B') = 0.9 \cdot 0.7 = 0.63 $$

### Решение на Python

```python
# Данные вероятности
P_A = 0.1  # Вероятность того, что первый телефон сломан
P_B = 0.3  # Вероятность того, что второй телефон сломан

# Вероятности работоспособности телефонов
P_A_working = 1 - P_A  # Вероятность, что первый телефон работает
P_B_working = 1 - P_B  # Вероятность, что второй телефон работает

# Вероятность того, что оба телефона работают (независимые события)
P_both_working = P_A_working * P_B_working

# Вывод результата
print(f"Вероятность того, что оба телефона работают: {P_both_working}")
```

# Вероятность получить ровно 4 орла при 6 бросках монеты

Монету бросают 6 раз. Какова вероятность того, что орел выпадет ровно 4 раза?

## Шаг 1: Определим условия задачи

- Количество бросков монеты (n): 6
- Количество успехов (орлов) (k): 4
- Вероятность успеха (выпадения орла) при одном броске (p): 0.5

## Шаг 2: Формула биномиального распределения

Вероятность того, что событие типа "успех" произойдёт ровно k раз за n независимых испытаний, определяется по формуле:

$$
P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}
$$

## Шаг 3: Подставим известные значения в формулу

Подставим $n = 6$, $k = 4$, $p = 0.5$ в формулу:

$$
P(X = 4) = \binom{6}{4} (0.5)^4 (0.5)^{6-4}
$$

## Шаг 4: Вычислим биномиальный коэффициент

Вычислим биномиальный коэффициент $\binom{6}{4}$:

$$
\binom{6}{4} = \frac{6!}{4!(6-4)!} = \frac{6!}{4!2!} = \frac{6 \times 5}{2 \times 1} = 15
$$

## Шаг 5: Вычислим вероятность

Подставим значение биномиального коэффициента в формулу:

$$
P(X = 4) = 15 \cdot (0.5)^4 \cdot (0.5)^{6-4} = 15 \cdot (0.5)^6
$$

Объединим степени:

$$
(0.5)^6 = \frac{1}{64}
$$

Теперь подставим это значение обратно:

$$
P(X = 4) = 15 \cdot \frac{1}{64} = \frac{15}{64}
$$

## Ответ

Таким образом, вероятность того, что орел выпадет ровно 4 раза при 6 бросках монеты, равна 

$$
\boxed{\frac{15}{64}}
$$


### Решение на Python

```python
import math

# Параметры задачи
n = 6  # Общее количество бросков
k = 4  # Количество орлов (успехов)
p = 0.5  # Вероятность успеха при одном броске

# Вычисляем биномиальный коэффициент
binomial_coefficient = math.comb(n, k)

# Вычисляем вероятность
probability = binomial_coefficient * (p ** k) * ((1 - p) ** (n - k))

# Выводим результат
print(f"Вероятность того, что орел выпадет ровно {k} раз при {n} бросках монеты: {probability:.4f}")
```

```python
# Импортируем необходимые библиотеки
import numpy as np

# Данные
a = 12  # длина нижнего основания
b = 27  # длина верхнего основания
alpha = 60  # угол в градусах

# Рассчитаем высоту h
h = (b - a) / 2 * np.tan(np.radians(alpha))

# Рассчитаем боковую сторону c
c = h / np.sin(np.radians(alpha))

# Периметр P
P = a + b + 2 * c

P
```
